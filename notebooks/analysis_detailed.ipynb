統計検定と仮説検証によるデータの深掘り分析

```python
import sys, pandas as pd, numpy as np
from pathlib import Path
sys.path.insert(0, str(Path('.').parent))
from src.hypothesis_verifier import HypothesisVerifier
print('✓ インポート完了')
```

## 1. データ準備と基本統計量

```python
def extract_numeric_data(file_path: str) -> list:
    df = pd.read_excel(file_path, sheet_name=0, header=None)
    df = df.iloc[4:].reset_index(drop=True)
    data_rows = []
    for idx, row in df.iterrows():
        label = str(row.iloc[0])
        values = [float(str(v).replace(',', '')) for v in row.iloc[1:] if pd.notna(v)]
        if values:
            data_rows.append({'label': label, 'values': values})
    return data_rows

file_path = Path('.').parent / 'data' / '第20表.xlsx'
data_rows = extract_numeric_data(str(file_path))
df = pd.DataFrame([{
    '名前': r['label'],
    '平均': np.mean(r['values']),
    '最大': np.max(r['values']),
    '最小': np.min(r['values'])
} for r in data_rows])
print(f'✓ {len(df)} データポイントをロード')
df.head()
```

## 2. 仮説検証

```python
verifier = HypothesisVerifier(alpha=0.05)
data = df['平均'].dropna().values
result1 = verifier.verify_normality(data, '平均値')
print(result1)
```

```python
result2 = verifier.verify_correlation(df['平均'].values, df['最大'].values, '平均', '最大')
print(result2)
```
